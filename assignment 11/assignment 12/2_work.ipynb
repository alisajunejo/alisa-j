```python
import pandas as pd 
```

```python
raw = pd.read_csv(r"C:\Users\hamza\Downloads\teaching_prices_messy (1).csv")
```

```python
#Check the problems
print(raw.info())
print("\nMissing values:")
print(raw.isna().sum())
print("\nDuplicates:")
print(raw.duplicated().sum())
```

```python
#Make a copy to clean 
df = raw.copy()

#clean coloumn names - remove spaces
df.columns = df.columns.str.strip().str.lower()

# 1) Missing value in 'price' - INVESTIGATE, do not delete
print("Rows with missing price:")
print(df[df['price'].isna()])

# 2) Duplicates - RETAIN, do not drop
# Duplicates are possible - 2 lessons can have the exact same price/date
# So we do NOT use drop_duplicates()

# 3) Fix 'price' - object -> numeric
# Remove $ , spaces, then convert
df['price'] = df['price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).str.strip()
df['price'] = pd.to_numeric(df['price'], errors='coerce')

# 4) Fix 'date' - object -> datetime
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# Final check
print(df.info())
print(df.head())
print(df.isna().sum())
```

```python
# BEFORE - save the evidence
print("=== BEFORE ===")
print(f"Rows: {len(raw)}")
print("\nMissing:")
print(raw.isna().sum())
print("\nDtypes:")
print(raw.dtypes)
print("\nPrice describe (as object):")
print(raw['price'].describe())
print("\nDate sample:")
print(raw['date'].head())
print("\nDuplicates:", raw.duplicated().sum())
```

```python
# AFTER - compare to BEFORE
print("=== AFTER ===")
print(f"Rows: {len(df)}")
print("\nMissing:")
print(df.isna().sum())
print("\nDtypes:")
print(df.dtypes)
print("\nPrice describe (as numeric):")
print(df['price'].describe())
print("\nDate sample:")
print(df['date'].head())
print("\nDuplicates:", df.duplicated().sum())
```

```python
# Did we remove any valid rows? Should be 0 removed
print(f"Rows lost: {len(raw) - len(df)}")

# Did any price value change without us knowing?
# Show rows where price became NaN after cleaning (should only be the 1 original missing)
print("\nPrices that became NaN after cleaning:")
print(df[df['price'].isna()])

# Does every item appear the number of times you expect?
# Replace 'item' with your actual column name if different, e.g. 'lesson' / 'subject'
print("\nCounts per item - BEFORE:")
print(raw.iloc[:, 0].value_counts())
print("\nCounts per item - AFTER:")
print(df.iloc[:, 0].value_counts())
```
